```python
import json
from pathlib import Path
```

```python
coco_path = Path(r"C:\dev\projects\CV_counting_bags\cvat_bags_coco\annotations\instances_default.json")
out_dir = Path(r"C:\dev\projects\CV_counting_bags\split_data")
out_dir.mkdir(exist_ok = True)

with open(coco_path, "r", encoding = "utf-8") as file:
    coco_data = json.load(file)

images = coco_data['images']
annotations = coco_data['annotations']
categories = coco_data['categories']
```

```python
train_images = [
    img for img in images
    if img['id'] < 225
]

val_images = [
    img for img in images
    if img['id'] >= 225
]

train_ids = {img['id'] for img in train_images}
val_ids = {img['id'] for img in val_images}

train_annotations = [
    ann for ann in annotations
    if ann['image_id'] in train_ids
]

val_annotations = [
    ann for ann in annotations
    if ann['image_id'] in val_ids
]
```

```python
train_coco = {
    "images": train_images,
    "annotations": train_annotations,
    "categories": categories
}

val_coco = {
    "images": val_images,
    "annotations": val_annotations,
    "categories": categories
}
```

```python
with open(out_dir / "instances_train.json", "w", encoding = "utf-8") as train_file:
    json.dump(train_coco, train_file, ensure_ascii = False, indent = 4)

with open(out_dir / "instances_val.json", "w", encoding = "utf-8") as val_file:
    json.dump(val_coco, val_file, ensure_ascii = False, indent = 4)
```
