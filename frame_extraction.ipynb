```python
import cv2
from pathlib import Path
```

```python
video_path = r"C:\dev\projects\CV_counting_bags\input.mp4"
out_dir = Path(r"C:\dev\projects\CV_counting_bags\frames_data")
out_dir.mkdir(exist_ok = True)

cap = cv2.VideoCapture(video_path)
fps = cap.get(cv2.CAP_PROP_FPS)
frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
duration = frames / fps

print("frames:", frames)
print("duration:", duration, "sec") 
```

```python
saved_frames = 0

for sec in range(0, int(duration), 2):
    cap.set(cv2.CAP_PROP_POS_MSEC, sec * 1000)
    ret, frame = cap.read()

    if ret:
        cv2.imwrite(str(out_dir / f"frame_{sec:04d}.jpg"), frame)
        saved_frames += 1

cap.release()
print("saved frames:", saved_frames)
```
